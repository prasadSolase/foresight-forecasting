```python
import pandas as pd
import numpy as np
```

```python
model_data = pd.read_csv("../data/cleaned/feature_engineered.csv")
```

```python
model_data["Date"] = pd.to_datetime(model_data["Date"])
```

```python
model_data.head()
```

```python
model_data = model_data.sort_values(
    ["Date", "SKU"]
).reset_index(drop=True)
```

```python
target = "Units_Sold"
```

```python
features = [
    "Price",
    "Promotion",
    "year",
    "month",
    "quarter",
    "week",
    "day_of_week",
    "is_weekend",
    "is_holiday",

    "Cost_Price",
    "Selling_Price",
    "Gross_Margin_Per_Unit",

    "Lag_1",
    "Lag_7",
    "Lag_14",
    "Lag_30",

    "Rolling_Mean_7",
    "Rolling_Mean_14",
    "Rolling_Mean_30",

    "Demand_Trend",

    "Current_Stock",
    "On_Order",
    "Lead_Time_Days",
    "Safety_Stock",
    "Reorder_Point",

    "Stock_Gap",
    "Below_Reorder_Point",
    "Below_Safety_Stock",
    "Stock_Coverage_Days"
]
```

```python
model_data[features].isnull().sum()
```

```python
model_data_ml = model_data.dropna(
    subset=features + [target]
).copy()
```

```python
print("Original rows:", len(model_data))
print("ML rows:", len(model_data_ml))
```

```python
X = model_data_ml[features]
y = model_data_ml[target]
```

```python
train_data = model_data_ml[
    model_data_ml["Date"] < "2025-01-01"
].copy()

test_data = model_data_ml[
    model_data_ml["Date"] >= "2025-01-01"
].copy()
```
